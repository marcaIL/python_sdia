Given the following variables, create and print the string

``quantity`` ``fruit``(s) cost(s) $``price``

such that
- there is an `"s"` at `"cost"` when there is more only one fruit,
- there is an `"s"` at `"fruit"` when there is more that one fruit,
- the price is displayed as ...,xxx,xxx.yy where yy corresponds rounded cents value

````python
str(round(1.8912392e4, 2))
````

````python
#add plurality if needed
def fruit_f(n,fruit_word):
    result=fruit_word,"cost"
    if (n>1):
        return result[0]+"s" +" "+ result[1]+"s"
    return result[0] +" "+ result[1]

#display numbers as asked : "xx,xxx,xxx.yy"
def format_price(price):
    int_part = int(price)
    str_frac = str(round(price,2)).split('.')[-1]
    converted = str(int_part)
    #On veut chercher à obtenir les entiers qui composent la partie entière
    #du prix 3 par 3. On teste d'abord si la taille de cette partie est >3
    size_first_part = len(converted)%3 or 3
    L = [converted[:size_first_part]]
    #Puis on va 3 par 3 sur les mots restants
    for i in range(size_first_part, len(converted),3):
        L.append(converted[i:i+3])
    result = ','.join(L)
    return f"{result}.{str_frac}"


def final_string(quantity, fruit, price):
    string = fruit_f(quantity, fruit) +" "+ format_price(price)
    return string
````

````python
#Basic unit test
import unittest

class TestFruitFunctions(unittest.TestCase):

    def test_final_string_singular(self):
        self.assertEqual(final_string(1, "apple", 1234.5693), "apple cost 1,234.57")

    def test_final_string_plural(self):
        self.assertEqual(final_string(3, "banana", 98765.4422), "bananas costs 98,765.44")

# Run tests
if __name__ == "__main__":
    unittest.main(argv=[''], exit=False)
````

Define a function `triangle_shape(height)`, that returns 
- `""` if `height=0`, 
- otherwise it returns a string that forms a triangle with height prescribed by `height`.

**Examples**

height=1
```
x
```

height=2
```
 x 
xxx
```

height=3
```
  x  
 xxx 
xxxxx
```

height=4
```
   x   
  xxx  
 xxxxx 
xxxxxxx
```

height=5
```
    x    
   xxx   
  xxxxx  
 xxxxxxx 
xxxxxxxxx
```

height=6
```
     x     
    xxx    
   xxxxx   
  xxxxxxx  
 xxxxxxxxx 
xxxxxxxxxxx
```

````python
def triangle_shape(height):
    list_odd = list(range(1, 2*height+1, 2))
    n = len(list_odd)-1

    string = ""

    if height>0 :
        height-=1
        for i in list_odd :
            line = (n-i//2)*" "+i*"x"+(n-i//2)*" "
            string += line+"\n"

    return string
````

````python
for i in range(7) :
    print(triangle_shape(i))
````
