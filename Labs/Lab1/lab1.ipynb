From the following list, create another list which contains
- `"fiz"` if the item is a multiple of 5,
- `"buz"` if the item is a multiple of 7,
- `"fizbuz"` if the item is a both a multiple of 5 and 7,
- the index of the item otherwise.

````python
x = [2, 1, 3, 31, 35, 20, 70, 132, 144, 49]
````

````python
def f(L,i):
    value = L[i]
    if (value%5==0 and value%7!=0):
        return "fiz"
    elif (value%7==0 and value%5!=0):
        return "buz"
    elif (value%7==0 and value%5==0):
        return "fizbuz"
    else:
        return i

result = [f(x,i) for i in range(len(x))]
````

Given the following variables, create and print the string

``quantity`` ``fruit``(s) cost(s) $``price``

such that
- there is an `"s"` at `"cost"` when there is more only one fruit,
- there is an `"s"` at `"fruit"` when there is more that one fruit,
- the price is displayed as ...,xxx,xxx.yy where yy corresponds rounded cents value

````python
quantity = 1
fruit = "avocado"
price = 1.8912392e4
````

````python
def fruit_f(n,fruit_word):
    result=fruit_word,"cost"
    if (n>1):
        return (result[0]+"s",result[1]+"s")
    return result
````

````python
def format_price(price):
    int_part = int(price)
    frac = price-int_part
    converted = str(int_part)
    #On veut chercher à obtenir les entiers qui composenet la partie entière
    #du prix 3 par 3. On teste d'abord si la taille de cette partie est >3
    size_first_part = len(converted)%3 or 3
    L = [converted[:size_first_part]]
    #Puis on va 3 par 3 sur les mots restants
    for i in range(size_first_part, len(converted),3):
        L.append(converted[i:i+3])
    result = ','.join(L)
    return f"{result}.{int(100*frac)}"
````

````python
result = fruit_f(quantity,fruit)
print(f"{quantity} {result[0]} {result[1]} ${format_price(price)}")
````

## Functions

Define the function `is_unique(x)` which takes a list of integers `x` as input and returns `True` if the there is no duplicate items otherwise return `False`.

````python
def is_unique(x):
    set_unique=set(x)
    n = len(set_unique)
    return (len(x)==n)
````

````python
print(is_unique([1,2,3]))
print(is_unique([2,2,3]))
````

````python
def triangle_shape(height):
    if (height<=0):
        return ""
    print("x"*height)
    return triangle_shape(height-1)
````
